```python
%cd /content/EasyEdit

import torch
print("torch:", torch.__version__, "| GPU:", torch.cuda.is_available())

from easyeditor import BaseEditor, ROMEHyperParams

hparams = ROMEHyperParams.from_hparams('./hparams/ROME/gpt2-xl.yaml')
hparams.model_name = 'gpt2-xl'      # 用 HF 仓库名，自动下载（yaml 里写的是本地路径）
hparams.mom2_n_samples = 5000

editor = BaseEditor.from_hparams(hparams)

prompts      = ['Question: What sport does Lionel Messi play? Answer:']
ground_truth = ['football']      # 原来的记忆
target_new   = ['basketball']    # 要改成的新记忆
subject      = ['Lionel Messi']

metrics, edited_model, _ = editor.edit(
    prompts=prompts,
    ground_truth=ground_truth,
    target_new=target_new,
    subject=subject,
    keep_original_weight=False,
)
print(metrics)
```

```python
%cd /content/EasyEdit
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

tok = AutoTokenizer.from_pretrained('gpt2-xl')
tok.pad_token = tok.eos_token

def gen(model, prompt, n=12):
    inp = tok(prompt, return_tensors='pt').to(model.device)
    with torch.no_grad():
        out = model.generate(**inp, max_new_tokens=n, do_sample=False,
                             pad_token_id=tok.eos_token_id)
    return tok.decode(out[0], skip_special_tokens=True)

p = 'Question: What sport does Lionel Messi play? Answer:'

# 编辑后的模型还在显存里，先生成
print("【编辑后】", gen(edited_model, p))

# 原模型放在 CPU 上加载，避免两张 gpt2-xl 同时占 12GB 显存
orig = AutoModelForCausalLM.from_pretrained('gpt2-xl', device_map='cpu')
print("【编辑前】", gen(orig, p, n=8))   # CPU 生成会慢 20-40 秒，正常
```

```python
import torch
from transformers import AutoTokenizer

tok = AutoTokenizer.from_pretrained('gpt2-xl')

def next_probs(model, prompt, words=(" basketball", " Soccer", " football")):
    device = next(model.parameters()).device
    inp = tok(prompt, return_tensors='pt').to(device)
    with torch.no_grad():
        logits = model(**inp).logits[0, -1]
    p = torch.softmax(logits.float(), dim=-1)
    return {w.strip(): round(float(p[tok(w, add_special_tokens=False).input_ids[0]]), 5)
            for w in words}

prompt = "Question: What sport does Lionel Messi play? Answer:"
print("编辑后:", next_probs(edited_model, prompt))
print("编辑前:", next_probs(orig, prompt))
```

```python
%cd /content/EasyEdit

metrics, edited_model, _ = editor.edit(
    prompts=['Question: What sport does Lionel Messi play? Answer:'],
    ground_truth=['football'],
    target_new=['basketball'],
    subject=['Lionel Messi'],
    sequential_edit=True,        # ← 关键：跳过还原，保留编辑结果
)
print(metrics[0]['post']['rewrite_acc'])
```

```python
prompt = "Question: What sport does Lionel Messi play? Answer:"
print("编辑后:", next_probs(edited_model, prompt))
print("编辑前:", next_probs(orig, prompt))
```
